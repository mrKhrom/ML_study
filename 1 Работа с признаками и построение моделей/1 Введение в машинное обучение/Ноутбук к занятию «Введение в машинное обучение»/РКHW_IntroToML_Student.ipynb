## Часть 2. Линейная регрессия
Обучите линейную регрессию на [датасете](https://www.kaggle.com/datasets/kumarajarshi/life-expectancy-who) ВОЗ о продолжительности жизни.

### 1. Подготовка данных
1. Скачайте датасет.
2. Импортируйте необходимые библиотеки.
3. Найдите целевую переменную (`life expectancy`).
4. Выберите ряд переменных в качестве признаков.
5. Проверьте данные на наличие пропусков и удалите их, так как модель линейной регрессии не умеет работать с пропусками.

```python
# Импортируйте необходимые библиотеки
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import pandas as pd

# Загрузите датасет и подготовьте данные (пропуски, выбор признаков)
df = pd.read_csv('Life Expectancy Data.csv')

df.columns = df.columns.str.strip() #Избавляемся от лишних пробелов

df = df.dropna() #удаляем все строки с пропусками

print(f'Размер таблицы:{df.shape}')
print(f'Количество пропусков = {df.duplicated().sum()}')

y = df['Life expectancy']
X = df[['Alcohol','percentage expenditure', 'Hepatitis B', 'BMI', 'Total expenditure', 'Income composition of resources', 'GDP']]
```

### 2. Обучение модели и оценка
6. Обучите модель линейной регрессии.
7. Выведите коэффициенты модели.
8. Выведите intercept (константу).
9. Для визуализации прямой выберите один наиболее значимый признак и постройте график зависимости целевой переменной от него.

```python
# Обучите модель
model = LinearRegression() #создаем модель
model.fit(X, y) #обучаем по значениям целевой переменной и параметрам

# Выведите коэффициенты и intercept
a = model.coef_          
b = model.intercept_   
print(f'Коэффициенты равны: {np.round(a,4)}')
print(f'Свободный член равен: {b}')

# Визуализируйте зависимость для признака, с самым большим весом. У нас это признак с индексом [5] - Income composition of resources
line_x=np.linspace(
    df['Income composition of resources'].min(),
    df['Income composition of resources'].max(),
    100)
line_y=line_x*a[5]+b

plt.figure(figsize=(7, 4))
plt.scatter(df[['Income composition of resources']], y, s=25, label="Данные")
plt.plot(line_x, line_y, color='red', linewidth=2)
plt.xlabel('Индекс уровня доходов населения')
plt.ylabel('Ожидаемая продолжительность жизни')
plt.title('Зависимость ожидаемой продолжительности жизни от индекса уровня доходов населения')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### 3. Преобразование признаков (полиномизация)
