**Пример реальной задачи:**

Кейс: в игре с внутриигровыми покупками нужно разложить игроков на группы похожих, без заранее заданных ярлыков. По поведению (время в игре, частота заходов, прогресс, участие в ивентах, суммы и частота покупок) алгоритм сам находит типы вроде «новички», «онлайн каждый день, но не платят», «редкие киты», «ивент-хантеры».

Практическая значимость: разным группам — разный контент и офферы (кому удержание, кому премиум-скин, кому обучение). Проще планировать live-ops, баланс и маркетинг: не одна рассылка на всех, а несколько понятных сегментов.

**Описание работы алгоритма:**

Что предсказывает модель: номер или метку кластера — «ты похож на группу 2», а не цену и не «уйдёт/останется» из готового словаря ответов. Смысл групп потом дают люди (аналитики, гейм-дизайнеры).

Тип обучения: без учителя.

Почему: нет целевой переменной y вроде «правильный сегмент». Есть только признаки X. Алгоритм сам ищет структуру: кто на кого похож. Если бы сегменты уже были размечены вручную и мы учили «угадать сегмент» — это была бы классификация с учителем.

Этапы обучения (простыми шагами)
1. Собрать логи: заходы, время в игре, прогресс, покупки.
2. Собрать таблицу признаков X по каждому игроку (без готового ответа y).
3. Почистить данные: боты, читеры, выбросы; привести признаки к удобному масштабу.
4. Выбрать число групп и алгоритм кластеризации (например, k-means).
5. Запустить алгоритм: разложить игроков по кластерам.
6. Посмотреть состав групп и дать им понятные имена.
7. Проверить на здравый смысл: отличаются ли группы по донату, удержанию, стилю игры.
8. Использовать сегменты в игре (офферы, ивенты, сложность онбординга).

Функция потерь / оценка ошибки: «правильного y» нет, поэтому ошибку считают иначе.
Смотрят, насколько плотные и отделённые группы (игроки внутри кластера похожи, между кластерами — нет).
Если группы «размазаны» или почти одинаковые — разбиение плохое, меняют число кластеров, признаки или алгоритм.
Задача — найти устойчивые, полезные для бизнеса сегменты, а не угадать готовую метку.

## Часть 2. Линейная регрессия
Обучите линейную регрессию на [датасете](https://www.kaggle.com/datasets/kumarajarshi/life-expectancy-who) ВОЗ о продолжительности жизни.

### 1. Подготовка данных
1. Скачайте датасет.
2. Импортируйте необходимые библиотеки.
3. Найдите целевую переменную (`life expectancy`).
4. Выберите ряд переменных в качестве признаков.
5. Проверьте данные на наличие пропусков и удалите их, так как модель линейной регрессии не умеет работать с пропусками.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

dataset = pd.read_csv("Life Expectancy Data.csv")

# print(dataset.columns.tolist())
# print(dataset.shape)
# dataset.head

# Целевая переменная y — продолжительность жизни
y_col = "Life expectancy "

# Признаки X — числовые факторы, которые могут влиять на y
feature_cols = [
    "Adult Mortality",
    "Alcohol",
    " BMI ",
    "HIV/AIDS".replace("HIV", " HIV") if False else " HIV/AIDS",
    "GDP",
    "Schooling",
    "Income composition of resources",
    "Polio",
    "Diphtheria ",
    "Total expenditure",
]

# Оставляем только y и выбранные признаки
data = dataset[[y_col] + feature_cols].copy()

# Удаляем строки с пропусками (линейная регрессия может ошибкаться из-за NaN)
data = data.dropna()


# X — признаки, y — цель
X = data[feature_cols]
y = data[y_col]

print("X:", X)
print("y", y)
X.head()
```

### 2. Обучение модели и оценка
6. Обучите модель линейной регрессии.
7. Выведите коэффициенты модели.
8. Выведите intercept (константу).
9. Для визуализации прямой выберите один наиболее значимый признак и постройте график зависимости целевой переменной от него.

```python
model = LinearRegression()
model.fit(X, y)

# --- 1. Смысловой вывод: что на что влияет ---
print("Целевая переменная y — ожидаемая продолжительность жизни (Life expectancy), в годах.\n")

print(f"Базовое значение (intercept b) = {model.intercept_:.2f} лет")
print("Это «старт» прогноза, когда влияние признаков учтено через коэффициенты.\n")

print("Как признаки влияют на продолжительность жизни:")
for name, coef in zip(feature_cols, model.coef_):
    if coef > 0:
        direction = "увеличивает"
    elif coef < 0:
        direction = "уменьшает"
    else:
        direction = "почти не меняет"
    print(f"  • {name.strip()}: коэффициент {coef:+.4f} - при росте признака на 1 "
          f"прогноз жизни в среднем {direction} примерно на {abs(coef):.4f} года")

print("\nКраткий вывод:")
print("  Сильнее всего обычно тянут вверх образование и доходный индекс,")
print("  вниз — HIV/AIDS и взрослая смертность.")

# --- 2. Прогноз для одного примера (значения можно менять вручную) ---

one_person = {
    "Adult Mortality": 120,              # смертность взрослых (меньше — обычно лучше)
    "Alcohol": 0.0,                      # потребление алкоголя
    " BMI ": 25.0,                       # индекс массы тела
    " HIV/AIDS": 0.1,                    # показатель HIV/AIDS
    "GDP": 5000,                         # ВВП на душу
    "Schooling": 23.0,                   # годы обучения
    "Income composition of resources": 0.65,
    "Polio": 90,                         # охват вакцинацией
    "Diphtheria ": 90,
    "Total expenditure": 6.0,            # расходы на здравоохранение, %
}

# контроль: те же столбцы и тот же порядок, что у X
x_one = pd.DataFrame([one_person], columns=feature_cols)

pred = model.predict(x_one)[0]

print("\n--- Прогноз для одного примера ---")
print("Входные значения:")
for k, v in one_person.items():
    print(f"  {k.strip()}: {v}")
print(f"\nПредсказанная продолжительность жизни: {pred:.1f} лет")


best_idx = int(np.argmax(np.abs(model.coef_)))
best_feature = feature_cols[best_idx]
best_coef = model.coef_[best_idx]

print(f"\nДля графика выбран признак: {best_feature.strip()}")
print(f"Его коэффициент: {best_coef:+.4f}")

# Данные только по этому признаку
x_best = X[best_feature].values
y_vals = y.values if hasattr(y, "values") else np.asarray(y)


# --- 3. График
# Прямая recиссии «y от одного признака» (для наглядной линии на 2D-графике)
# Это отдельная простая модель: life expectancy ≈ a * feature + b
model_1d = LinearRegression()
model_1d.fit(x_best.reshape(-1, 1), y_vals)

x_line = np.linspace(x_best.min(), x_best.max(), 200).reshape(-1, 1)
y_line = model_1d.predict(x_line)

plt.figure(figsize=(8, 5))
plt.scatter(x_best, y_vals, s=15, alpha=0.35, label="Реальные данные")
plt.plot(x_line, y_line, color="red", linewidth=2, label="Прямая линейной регрессии")
plt.xlabel(best_feature.strip())
plt.ylabel("Life expectancy (годы)")
plt.title(f"Зависимость продолжительности жизни от «{best_feature.strip()}»")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print(
    f"Прямая на графике: y ≈ {model_1d.coef_[0]:+.4f} * ({best_feature.strip()}) "
    f"+ {model_1d.intercept_:.2f}"
)
```
